```python
# Funkcja do ekstrakcji tekstu z PDF
def extract_text_from_pdf(pdf_path, method='pdfplumber'):
    """
    Wyciąga tekst z pliku PDF.
    
    Args:
        pdf_path (str): Ścieżka do pliku PDF
        method (str): 'pdfplumber' lub 'pypdf2'
    
    Returns:
        str: Wyekstraktowany tekst
    """
    
    if method == 'pdfplumber':
        with pdfplumber.open(pdf_path) as pdf:
            text = ""
            for page in pdf.pages:
                text += page.extract_text()
            return text
    
    elif method == 'pypdf2':
        with open(pdf_path, 'rb') as file:
            pdf_reader = PyPDF2.PdfReader(file)
            text = ""
            for page in pdf_reader.pages:
                text += page.extract_text()
            return text
    
    else:
        raise ValueError(f"Nieznana metoda: {method}. Użyj 'pdfplumber' lub 'pypdf2'")


# Test funkcji
print("🧪 TEST FUNKCJI:")
print("=" * 50)

# Test z pdfplumber
result = extract_text_from_pdf(pdf_path, method='pdfplumber')
print(f"✅ pdfplumber: {len(result)} znaków")

# Test z PyPDF2
result = extract_text_from_pdf(pdf_path, method='pypdf2')
print(f"✅ PyPDF2: {len(result)} znaków")

print("\n🎉 Funkcja działa poprawnie!")
```

```python
# ULEPSZONA funkcja z obsługą błędów
def extract_text_from_pdf(pdf_path, method='pdfplumber', clean=True):
    """
    Wyciąga tekst z pliku PDF z obsługą błędów.
    
    Args:
        pdf_path (str): Ścieżka do pliku PDF
        method (str): 'pdfplumber' lub 'pypdf2'
        clean (bool): Czyczyścić tekst (usunąć nadmiarowe spacje/newliny)
    
    Returns:
        str: Wyekstraktowany tekst lub None w przypadku błędu
    """
    
    try:
        # Sprawdź czy plik istnieje
        if not os.path.exists(pdf_path):
            print(f"❌ Błąd: Plik nie istnieje: {pdf_path}")
            return None
        
        # Ekstrakcja
        if method == 'pdfplumber':
            with pdfplumber.open(pdf_path) as pdf:
                text = ""
                for page in pdf.pages:
                    page_text = page.extract_text()
                    if page_text:  # Sprawdź czy strona ma tekst
                        text += page_text + "\n"
        
        elif method == 'pypdf2':
            with open(pdf_path, 'rb') as file:
                pdf_reader = PyPDF2.PdfReader(file)
                text = ""
                for page in pdf_reader.pages:
                    page_text = page.extract_text()
                    if page_text:
                        text += page_text + "\n"
        
        else:
            print(f"❌ Błąd: Nieznana metoda '{method}'")
            return None
        
        # Czyszczenie tekstu (opcjonalne)
        if clean and text:
            # Usuń nadmiarowe spacje
            text = ' '.join(text.split())
        
        return text
    
    except Exception as e:
        print(f"❌ Błąd podczas ekstrakcji: {e}")
        return None


# Test ulepszonej funkcji
print("🧪 TEST ULEPSZONEJ FUNKCJI:")
print("=" * 50)

result = extract_text_from_pdf(pdf_path, method='pdfplumber', clean=True)
if result:
    print(f"✅ Sukces! Długość: {len(result)} znaków")
    print(f"📄 Fragment: {result[:200]}...")
else:
    print("❌ Ekstrakcja nie powiodła się")
```

```python
# Funkcja porównująca obie metody
def compare_pdf_methods(pdf_path):
    """
    Porównuje PyPDF2 i pdfplumber dla danego PDF.
    Zwraca statystyki i rekomendację.
    
    Args:
        pdf_path (str): Ścieżka do pliku PDF
    
    Returns:
        dict: Wyniki porównania
    """
    
    print("🔍 PORÓWNANIE METOD EKSTRAKCJI")
    print("=" * 60)
    
    # Test pdfplumber
    text_pdfplumber = extract_text_from_pdf(pdf_path, method='pdfplumber', clean=True)
    
    # Test PyPDF2
    text_pypdf2 = extract_text_from_pdf(pdf_path, method='pypdf2', clean=True)
    
    if not text_pdfplumber or not text_pypdf2:
        print("❌ Nie udało się wyekstraktować tekstu")
        return None
    
    # Statystyki
    results = {
        'pdfplumber': {
            'length': len(text_pdfplumber),
            'text': text_pdfplumber
        },
        'pypdf2': {
            'length': len(text_pypdf2),
            'text': text_pypdf2
        }
    }
    
    # Wyświetl porównanie
    print(f"\n📊 STATYSTYKI:")
    print(f"{'Metoda':<15} {'Długość':<10} {'Ocena'}")
    print("-" * 60)
    print(f"{'pdfplumber':<15} {results['pdfplumber']['length']:<10} {'✅ Zwykle lepsze formatowanie'}")
    print(f"{'PyPDF2':<15} {results['pypdf2']['length']:<10} {'⚠️ Prostsze, szybsze'}")
    
    # Rekomendacja
    print(f"\n💡 REKOMENDACJA:")
    if abs(results['pdfplumber']['length'] - results['pypdf2']['length']) < 100:
        print("   Obie metody dają podobne wyniki - wybierz dowolną")
        results['recommended'] = 'both'
    else:
        print("   pdfplumber - lepsze formatowanie dla złożonych PDF")
        results['recommended'] = 'pdfplumber'
    
    return results


# Test funkcji porównawczej
comparison = compare_pdf_methods(pdf_path)
```
